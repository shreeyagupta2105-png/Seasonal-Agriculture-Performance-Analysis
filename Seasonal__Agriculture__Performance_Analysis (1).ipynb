# Data Cleaning and Preparation

Before performing analysis, the dataset needs to be checked for missing values, duplicate records, and data quality issues.

```python
# Missing Values

df.isnull().sum().sort_values(ascending=False)
```

## Missing Value Analysis

The dataset contains a small number of missing values in three numerical columns:

- Rainfall_mm
- Soil_Moisture_pct
- Yield_Tonnes_Ha

Since the percentage of missing values is very low compared to the overall dataset size, removing records could lead to unnecessary loss of information. Therefore, missing values will be replaced using the median value of the respective columns.

```python
# Duplicate Records

df.duplicated().sum()
```

##Handle Missing Values

```python
# Filling Missing Values

df['Rainfall_mm'].fillna(df['Rainfall_mm'].median(), inplace=True)

df['Soil_Moisture_pct'].fillna(
    df['Soil_Moisture_pct'].median(),
    inplace=True
)

df['Yield_Tonnes_Ha'].fillna(
    df['Yield_Tonnes_Ha'].median(),
    inplace=True
)
```

```python
df.isnull().sum().sum()
```

### Data Cleaning Summary

- No major data quality issues were identified.
- Missing values were present in three numerical variables.
- Median imputation was used to handle missing values.
- The dataset is now complete and ready for analysis.

# Data Visualization and Seasonal Performance Analysis

The primary objective of this project is to understand how agricultural performance varies across seasons.

This section uses data visualization techniques to compare important agricultural indicators such as yield, production, profit, water usage, and disease risk across different seasons. The analysis helps identify seasonal patterns and factors that influence agricultural outcomes.

## **Research Question 1**

### How does agricultural yield vary across seasons?

Yield is one of the most important indicators of agricultural performance. Comparing average yield across seasons helps identify which season supports better crop productivity.

```python
season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False)

plt.figure(figsize=(7,5))

sns.barplot(
    x=season_yield.index,
    y=season_yield.values
)

plt.title('Average Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')

plt.show()

season_yield
```

### Observation

The **Kharif** season recorded the highest average yield, while **Zaid** recorded the lowest.

### Insight

Seasonal conditions appear to influence crop productivity. Variations in rainfall, temperature, and soil moisture may contribute to differences in yield across seasons.

### Recommendation

Agricultural planning should focus on crops that perform consistently well during high-yield seasons.

## **Research Question 2**

### Which season generates the highest profit?

Profitability is an important indicator because high production does not always result in high profit. This analysis compares average profit across seasons.

```python
season_profit = (
    df.groupby('Season')['Profit_INR']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(7,5))

sns.barplot(
    x=season_profit.index,
    y=season_profit.values
)

plt.title('Average Profit by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')

plt.show()

season_profit
```

### Observation
Kharif season generated the highest average profit (₹178,914.65), followed by Rabi (₹87,689.47). Zaid recorded a negative average profit (-₹24,804.82), indicating losses during this season.
### Insight
The results indicate that Kharif is the most profitable season in the dataset. In contrast, the negative profit observed during Zaid suggests that production costs may exceed revenue for many farms during this season.

### Recommendation
Farmers should prioritize profitable crop cultivation during the Kharif season while carefully evaluating input costs and crop selection strategies during the Zaid season.

## **Research Question 3**

### Which crops perform best in different seasons?

Different crops respond differently to seasonal conditions. This analysis identifies crop-season combinations associated with higher productivity.

```python
crop_season = pd.pivot_table(
    df,
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

plt.figure(figsize=(10,6))

sns.heatmap(
    crop_season,
    annot=True,
    fmt='.2f',
    cmap='YlGnBu'
)

plt.title('Average Yield by Crop and Season')

plt.show()
```

### Observation

Yield performance varies across crops and seasons. Certain crop-season combinations consistently produce higher yields than others.

### Insight

Crop selection should be aligned with seasonal conditions rather than using a uniform approach throughout the year.

### Recommendation

Farmers should prioritize crop-season combinations that demonstrate strong productivity.

## **Research Question 5**

## **Research Question 6**

### Which irrigation method provides better water efficiency?

Efficient water management is crucial for sustainable agriculture. This analysis compares different irrigation methods to determine which approach delivers better water efficiency and resource utilization.

```python
irrigation_eff = (
    df.groupby('Irrigation_Method')
      ['Water_Efficiency_t_per_1000m3']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

sns.barplot(
    x=irrigation_eff.index,
    y=irrigation_eff.values
)

plt.title('Average Water Efficiency by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Water Efficiency')

plt.xticks(rotation=15)

plt.show()

irrigation_eff
```

### Observation
Rainfed irrigation achieved the highest average water efficiency (7.56 t per 1000 m³), followed by Drip irrigation (6.27). Flood irrigation recorded the lowest water efficiency (3.44).

### Insight
The analysis suggests that Rainfed and Drip irrigation methods utilize water more efficiently than Sprinkler and Flood irrigation methods. Flood irrigation appears to require significantly more water for similar levels of output.

### Recommendation
Farmers should consider adopting water-efficient irrigation practices such as Drip irrigation wherever feasible. Reducing dependence on Flood irrigation can help improve water conservation and agricultural sustainability.

## **Research Question 7**

### Which environmental factors have the strongest impact on yield and profit?

Agricultural performance is influenced by multiple environmental conditions such as rainfall, temperature, humidity, soil moisture, and soil characteristics. This analysis investigates which factors show the strongest relationship with crop yield and profitability.

```python
corr_cols = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Sunlight_Hours_Day',
    'Soil_pH',
    'Soil_Moisture_pct',
    'Yield_Tonnes_Ha',
    'Profit_INR'
]

plt.figure(figsize=(10,6))

sns.heatmap(
    df[corr_cols].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
yield_corr = df[corr_cols].corr()['Yield_Tonnes_Ha'].sort_values(ascending=False)
yield_corr
```

```python
profit_corr = df[corr_cols].corr()['Profit_INR'].sort_values(ascending=False)
profit_corr
```

### Additional Finding

Rainfall and soil moisture show stronger positive relationships with yield compared to temperature and soil pH.

### Observation
Yield and Profit show a moderate positive correlation (0.49), indicating that higher yields generally contribute to higher profitability.

Among environmental variables, Rainfall and Soil Moisture exhibit moderate positive relationships with each other (0.51), while Humidity is also positively associated with Rainfall (0.52). Temperature and Soil pH show very weak relationships with Yield and Profit.

### Insight
The findings suggest that water-related factors such as rainfall, humidity, and soil moisture play a more important role in agricultural performance than temperature or soil pH within this dataset. However, no single environmental variable shows a strong direct correlation with yield, indicating that agricultural productivity is influenced by multiple interacting factors.
### Recommendation
Farmers should focus on maintaining adequate soil moisture and effective water management practices. Monitoring rainfall patterns and soil moisture levels can support better agricultural planning and resource allocation.

## **Research Question 8**

### Which states show the best agricultural performance?

Agricultural outcomes can vary across regions due to differences in climate, soil conditions, farming practices, and resource availability. This analysis compares agricultural performance across states to identify high-performing regions.

### Observation

Agricultural performance varies across states, with some states recording higher average yields than others.

### Insight

Differences in climate, soil conditions, irrigation infrastructure, and farming practices may contribute to variations in productivity among states.

### Recommendation

States demonstrating strong agricultural performance can serve as useful references for identifying effective farming practices and resource management strategies.

```python
state_profit = (
    df.groupby('State')['Profit_INR']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=state_profit.values,
    y=state_profit.index
)

plt.title('Average Profit by State')
plt.xlabel('Average Profit (INR)')
plt.ylabel('State')

plt.show()
```
